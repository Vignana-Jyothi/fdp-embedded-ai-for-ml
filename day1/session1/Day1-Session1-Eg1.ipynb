```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

    
```

```python

def extract_channel(tiff_path, channel):
    # Read TIFF image
    img = Image.open(tiff_path).convert("RGB")

    # Convert to NumPy array
    arr = np.array(img)

    # Create black image
    result = np.zeros_like(arr)

    # Channel index: R=0, G=1, B=2
    channel_index = {
        "red": 0,
        "green": 1,
        "blue": 2
    }[channel.lower()]

    # Keep only the selected channel
    result[:, :, channel_index] = arr[:, :, channel_index]

    return result

```

```python

# ---------------------------------------------------
# Change this to your TIFF file
# ---------------------------------------------------
tiff_file = "input.tiff"

# Choose: "red", "green", or "blue"
channel = "green"

# Extract channel
result = extract_channel(tiff_file, channel)

# Display
plt.figure(figsize=(10, 8))
plt.imshow(result)
plt.title(f"{channel.capitalize()} Channel")
plt.axis("off")
plt.show()
```
