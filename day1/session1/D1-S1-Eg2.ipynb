```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

tiff_file = "input.tiff"

img = Image.open(tiff_file).convert("RGB")
arr = np.array(img)

channels = {
    "Red": 0,
    "Green": 1,
    "Blue": 2
}

for name, index in channels.items():

    result = np.zeros_like(arr)
    result[:, :, index] = arr[:, :, index]

    plt.figure(figsize=(10, 8))
    plt.imshow(result)
    plt.title(f"{name} Channel")
    plt.axis("off")
    plt.show()
```
