```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt


def extract_channel(tiff_path, channel):

    # Read TIFF image
    img = Image.open(tiff_path).convert("RGB")

    # Convert to NumPy array
    arr = np.array(img)

    channel_index = {
        "red": 0,
        "green": 1,
        "blue": 2
    }[channel.lower()]

    # Extract only the selected channel
    result = arr[:, :, channel_index]

    return result


# ---------------------------------------------------
# TIFF file
# ---------------------------------------------------
tiff_file = "input.tiff"

# Select: "red", "green", or "blue"
channel = "green"

# Extract channel
result = extract_channel(tiff_file, channel)


# ---------------------------------------------------
# Display
# ---------------------------------------------------
plt.figure(figsize=(10, 8))

plt.imshow(result, cmap="gray")

plt.title(f"{channel.capitalize()} Channel")
plt.colorbar()

plt.axis("off")
plt.show()
```

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt


tiff_file = "input.tiff"
channel = "green"

img = Image.open(tiff_file).convert("RGB")
arr = np.array(img)

channel_index = {
    "red": 0,
    "green": 1,
    "blue": 2
}[channel.lower()]

# Create black image
result = np.zeros_like(arr)

# Put selected channel back
result[:, :, channel_index] = arr[:, :, channel_index]


plt.figure(figsize=(10, 8))
plt.imshow(result)
plt.title(f"{channel.capitalize()} Channel")
plt.axis("off")
plt.show()
```
